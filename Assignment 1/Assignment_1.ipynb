### Part No 2:

Using the Glasglow Open Data API ( Transit) https://developer.glasgow.gov.uk/api-details#api=traffic&operation=traffic-sensor-locations fetch all the sensor locations in the city. Find the WorkingZones and Calculate/Map the areas with more and fewer sensors distributed in the city.


#### Sensor Data Preperation

```python
import requests
import pandas as pd
import geopandas as gpd
import leafmap

# best to link URL like this, so in the future, you can easily update the URL
glasgow_sensors = "https://api.glasgow.gov.uk/traffic/v1/movement/sites"

# Making the query to the web server, using the Get method from the requests library 
response = requests.get(glasgow_sensors)
data = response.json()
```

```python
# empty lists to store what we need
site_ids = []
from_lats = []
from_longs = []
from_descriptions = []

# Iterate over each entry in the JSON array
for entry in data:
    site_id = entry['siteId']
    from_description = entry['from']['description']
    from_lat = entry['from']['lat']
    from_long = entry['from']['long']
    
    # Append values to lists
    site_ids.append(site_id)
    from_descriptions.append(from_description)
    from_lats.append(from_lat)
    from_longs.append(from_long)
    
# Create a Pandas DataFrame, using the lists.
glasgow_sensors = pd.DataFrame({
    'siteId': site_ids,
    'from_description': from_descriptions,
    'from_lat': from_lats,
    'from_long': from_longs,
})
```

```python
glasgow_sensors.dtypes
```

```python
# convert latitude and longitude from text to number
glasgow_sensors['from_lat'] = glasgow_sensors['from_lat'].astype(float)
glasgow_sensors['from_long'] = glasgow_sensors['from_long'].astype(float)
```

```python
glasgow_sensors.head()
```

```python
# filtering to get rid of records without latitude or longitude values 
glasgow_sensors = glasgow_sensors[
    (glasgow_sensors['from_lat'].notna()) &
    (glasgow_sensors['from_long'].notna()) &
    (glasgow_sensors['from_lat'] != 0) & 
    (glasgow_sensors['from_long'] != 0)
]

# how many rows and columns are left
print(glasgow_sensors.shape)
```

```python
# creating a geodataframe and setting CRS
glasgow_sensors_gdf = gpd.GeoDataFrame(glasgow_sensors, geometry=gpd.points_from_xy(glasgow_sensors['from_long'], glasgow_sensors['from_lat']), crs='EPSG:4326')

#investigating geodataframe
glasgow_sensors_gdf.head()
```

#### Working Zone Data Preparation

```python
# importing working zones data
all_working_zones_gdf = gpd.read_file("Assignment 1/Assignment_1_data/WorkplaceZones2011Scotland/WorkplaceZones2011Scotland.shp")

#investigating working zone geodataframe
all_working_zones_gdf.head()
```

```python
# setting CRS
all_working_zones_gdf = all_working_zones_gdf.to_crs("EPSG:4326")
```

```python
# filtering to create a geodataframe with on Glasgow working zones
only_glasgow_working_zones = all_working_zones_gdf[all_working_zones_gdf["LADCD"] == "S12000046"]
only_glasgow_working_zones = gpd.GeoDataFrame(only_glasgow_working_zones)
only_glasgow_working_zones.head()
```

#### Merging Data

```python
# Using sJoin to calculate the overlay of sensors and polygons
working_zone_with_sensors = gpd.GeoDataFrame(gpd.sjoin(glasgow_sensors_gdf, all_working_zones_gdf, how = "inner", predicate = "intersects", lsuffix = "left", rsuffix = "right"))
working_zone_with_sensors.head()
```

```python
sensors_zone_cols = gpd.GeoDataFrame(working_zone_with_sensors[["siteId", "WZCD"]])

#Using group_by to count the number of sensors per WorkingZone
sensors_per_zone = sensors_zone_cols.groupby("WZCD").count()
sensors_per_zone = sensors_per_zone.rename(columns={"siteId": "sensor_count"}).reset_index()
sensors_per_zone.head()
```

```python
# merge the geodataframes together, keeping every Glasgow working zone regardless of wether it has sensor data associated with it
zone_counts = only_glasgow_working_zones.merge(sensors_per_zone, on="WZCD", how="left")
```

#### Map of Sensor Distrubtion Across Glasgow's Working Zones

```python
zone_counts.explore("sensor_count", cmap = "Oranges")
```

### The differences between working with Longboard and Datashader

Longboard focuses on interactive visualisation of large datasets through maps and charts, whereas Datashader focuses on efficiently rendering large amounts of data points to reveal patterns and reduce visual clutter.

Lonboard is more exciting for interactive exploration, as it allows you to investigate individual locations and customise map layers. Datashader creates a static and aggregated visualisation that makes overall patterns and concentrations easier to identify with large datasets.

### Data Preparation

```python
import zipfile
import os
import pandas as pd

zip_path = "Assignment 1/Assignment_1_data/dft_traffic_counts_raw_counts.zip"
extract_path = "Assignment 1/Assignment_1_data/traffic_counts_extracted"

# Unzip the file. YOU NEED TO UPDATE THIS PATH based on where you place the zip file.
with zipfile.ZipFile(zip_path, 'r') as zip_ref:
    zip_ref.extractall(extract_path)

# List all .csv files in subdirectories
# Here we use https://docs.python.org/dev/library/pathlib.html and walk https://docs.python.org/dev/library/os.html#os.walk
csv_files = [os.path.join(root, file) for root, dirs, files in os.walk(extract_path) for file in files if file.endswith('.csv')]

traffic_df = pd.concat([pd.read_csv(csv_file) for csv_file in csv_files], ignore_index=True)
```

```python
# seeing if the dataset has at least 5 million records
traffic_df.shape
```

```python
# investigating what information each record holds
traffic_df.columns
```

```python
traffic_df.head()
```

```python
# checking to see if numeric values have the right data type
traffic_df[
    ["latitude", "longitude", "all_motor_vehicles"]
].info()
```

```python
# checking to see if there's any missing values in records
traffic_df[
    ["latitude", "longitude", "all_motor_vehicles"]
].isna().any()
```

```python
# getting rid off records with missing values
clean_traffic_df = traffic_df.dropna(subset=["all_motor_vehicles"])
```

```python
# filtering records (taking advice of only using a small portion of that rather than work with the entire table)
traffic_2024 = clean_traffic_df[clean_traffic_df["year"] == 2024]

traffic_2024.tail()
```

```python
# checking to see how many records there are after filtering
traffic_2024.shape
```

```python
# checking to see if count_point_id all have the same coordinates - 1 max unique value means that each count_point_id has only one longitude and latitude associated with it 
traffic_2024.groupby("count_point_id")[["longitude", "latitude"]].nunique().max()
```

```python
# checking how many count point IDs there are
clean_traffic_df["count_point_id"].nunique()
```

```python
# creating single row per count_point_id (location) that shows the average recorded traffic count across all days, hours and directions
mean_traffic_by_point_2024 = (
    traffic_2024.groupby("count_point_id", as_index=False)
    .agg(longitude=("longitude", "first"), 
        latitude=("latitude", "first"),
        mean_traffic=("all_motor_vehicles", "mean"),
        region_id=("region_id", "first"),
        region_name=("region_name", "first"),
        road_name=("road_name", "first"),
        road_category=("road_category", "first"),
        road_type=("road_type", "first")))

mean_traffic_by_point_2024.head()
```
